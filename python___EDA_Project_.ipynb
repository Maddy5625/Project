Data visualisation

Histogram

```python
population_cols = ['Population_Millions', 'Median_Age', 'Birth_Rate_per_1000']

df[population_cols].hist(
    figsize=(12, 6),
    bins=10,
    edgecolor='black'
)

plt.suptitle("Histogram - Population Overview")
plt.tight_layout()
plt.show()
```

Bar chart

```python
df.groupby('Country')['Population_Millions'].mean().sort_values(ascending=False).plot(kind='bar', edgecolor='black')
plt.title("Average Population by Country")
plt.xlabel("Country")
plt.ylabel("Population")
plt.xticks(rotation=45)
plt.show()
```

Box plot

```python
sns.boxplot(data=df, y="Median_Age")
plt.title("Median Age Outliers")
plt.show()
```

Scatter plot

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='Median_Age',
    y='Birth_Rate_per_1000',
    hue='Country',
    s=100
)
plt.title("Scatter Plot: Birth Rate vs. Median Age")
plt.xlabel("Median Age")
plt.ylabel("Birth Rate (per 1000)")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

 Heatmap

```python
sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()
```

 Pie chart

```python
import matplotlib.pyplot as plt

# Calculate total population per country in our demographic dataset
country_populations = df.groupby('Country')['Population_Millions'].sum()

plt.figure(figsize=(8, 8))
plt.pie(
    country_populations,
    labels=country_populations.index,
    autopct='%1.1f%%',
    startangle=140
)

plt.title("Population of Country")
plt.show()
```
